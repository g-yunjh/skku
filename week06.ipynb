## d. 잠깐! 옆으로 나와서... matplotlib 플롯의 구조 🛟
---
- matplotlib = 플롯은 Figure 객체(도화지), Axes 객체(그림), Axis 객체(그림에 포함된 x, y축)로 구성
- 따라서, 당연히 모든 플롯은 `Figure 객체` = 정확하게는 matplotlib.figure.Figure 클래스 객체에 포함 --> 도화지 한 장당 윈도우 하나 배정
- 원래는 figure 명령을 사용하여 Figure 객체를 생성 --> 우리 figure 사용 안했잖아요. --> plt 서브패키지가 자동으로 Figure, AX를 생성
- http://matplotlib.org/api/figure_api.html#matplotlib.figure.Figure<br><br>

<img src='https://raw.githubusercontent.com/tysep16/DV25_1/refs/heads/main/figaxesaxis.png' width=600>

- plt.figure() 함수의 매개변수 = plt.figure(매개변수)
    - num : 그림의 고유 번호 또는 이름. 이 매개변수를 사용하면 나중에 그림을 참조
    - figsize : 그림의 너비와 높이를 지정하는 튜플. 단위는 인치.
    - dpi : 그림의 해상도를 지정하는 정수값. "dots per inch"의 약자.
    - facecolor : 그림의 배경색을 지정하는 색상값.
    - edgecolor : 그림의 테두리 색을 지정하는 색상값.
    - frameon : 그림의 프레임을 그릴지 여부를 지정하는 불린 값.

```python
t = np.arange(0, 5, 0.2)
Fig = plt.figure(figsize=(10, 2)) # 도화지 크기 설정
plt.plot(t, t, 'r--', label='test1')
plt.plot(t, t**2, 'bs:', label='test2')
plt.plot(t, t**3, 'g^-', label='test3')
plt.legend(loc=2)
plt.tight_layout()
plt.show()
```

- `Axes 객체` --> Figure 안에 있는 Axes 객체를 여러 개 생성 = subplot로 지정
- 자세한 설명 : http://matplotlib.org/api/axes_api.html#matplotlib.axes.Axes
- 하면서 배워 보자.

```python
t = np.arange(0, 5, 0.2)
Fig = plt.figure(figsize=(10, 5))
Ax1 = plt.subplot(2, 1, 1) # Ax1 = Fig.add_subplot(2, 1, 1)
plt.plot(t, t, 'r--', label='test1') # Ax1.plot(t, t, 'r--', label='test1')
Ax2 = plt.subplot(2, 1, 2)
plt.plot(t, t**2, 'bs:', label='test2')
plt.plot(t, t**3, 'g^-', label='test3')
plt.legend(loc=2) # 변경해보자
plt.tight_layout()
plt.show()
```

- 이 때 subplot 의 인수는 (2,2,1)를 줄여서 221 라는 하나의 숫자로 표시 가능 --> **Axes의 위치는 위에서 부터 아래로, 왼쪽에서 오른쪽으로 카운트**

```python
plt.subplot(221)
plt.plot(np.random.rand(5))
plt.title("axes 1")

plt.subplot(222)
plt.plot(np.random.rand(5))
plt.title("axes 2")

plt.subplot(223)
plt.plot(np.random.rand(5))
plt.title("axes 3")

plt.subplot(224)
plt.plot(np.random.rand(5))
plt.title("axes 4")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 400)
y = np.sin(x ** 2)
 
fig, axs = plt.subplots(2, 2)
# fig = plt.figure()
# axs = fig.subplots(2, 2)
 
axs[0, 0].plot(x, y)
axs[0, 0].set_title('Axis [0, 0]')
 
axs[0, 1].plot(x, y, c='orange')
axs[0, 1].set_title('Axis [0, 1]')
 
axs[1, 0].plot(x, -y, c='green')
axs[1, 0].set_title('Axis [1, 0]')
 
axs[1, 1].plot(x, -y, c='red')
axs[1, 1].set_title('Axis [1, 1]')

plt.tight_layout()
plt.show()
```

- 정리하면.
- subplot()은 Figure를 암묵적으로 만들지만,
- subplots()은 Figure와 Axes를 명시적으로 다룰 수 있어 더 강력!!<br><br>


| 항목            | `plt.subplot()`                      | `plt.subplots()`                                      |
|-----------------|---------------------------------------|--------------------------------------------------------|
| **기본 목적**   | 단일 subplot 생성                    | 여러 subplot을 한 번에 생성                            |
| **리턴값**      | `Axes` 객체 하나                     | `(Figure, Axes)` 튜플 반환                             |
| **유연성**      | 낮음 – 하나씩 추가해야 함            | 높음 – 전체 구조를 미리 만들고 조작 가능              |
| **Figure 제어** | 불가능 (암묵적으로 생성됨)           | 가능 (명시적으로 `Figure` 객체 반환됨)                |
| **추천 여부**   | 옛날 방식 / 간단한 경우에만 사용     | 현대적 / 대부분의 경우에 추천 

- `Axis 객체` --> 하나의 Axes 객체 안에는 두 개 이상의 Axis 객체 --> 복수의 y 축 = twinx 명령
- https://matplotlib.org/api/axis_api.html<br><br>

<img src='https://raw.githubusercontent.com/tysep16/DV25_1/refs/heads/main/figures.png' width=600>

```python
fig, ax0 = plt.subplots()
ax1 = ax0.twinx()
ax0.set_title("example of twinx")
ax0.plot([10, 5, 2, 9, 7], 'r-', label="y0")
ax0.set_ylabel("y0")
ax0.grid(False)
ax1.plot([100, 200, 220, 180, 120], 'g:', label="y1")
ax1.set_ylabel("y1")
ax1.grid(False)
ax0.set_xlabel("x axis")
plt.show()
```

## e. 다시 돌아와서, matplotlib --> bar chart 🍹
---
- x 데이터가 카테고리 값인 경우 = y 데이터는 스칼라 값인 경우 --> bar/barh = 바 차트(bar chart) = plt.bar(파라미터)
- http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.bar
- http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.barh

- 설정 가능한 파라미터는?
    - x : x 축에 위치할 데이터의 일련의 값
    - height : 각 막대의 높이를 나타내는 데이터의 일련의 값
    - width : (선택) 각 막대의 너비를 설정. 디폴트 값은 0.8
    - bottom : (선택) 각 막대의 y 축에서의 위치를 설정. 디폴트 값은 기본 위치인 y=0
    - align : 막대의 x 좌표를 막대의 가운데로 할 것인지, 막대의 왼쪽 끝으로 할 것인지를 결정. 디폴트 값은 'center'
    - color : 막대의 색
    - edgecolor : 막대의 테두리 색
    - linewidth : 막대의 테두리 너비
    - tick_label : 각 막대에 대한 레이블
    - label : 범례에 사용될 레이블

```python
y = [2, 3, 1]
x = range(3)
# x = np.arange(len(y))

plt.title("Bar Chart")
plt.bar(x, y) # 옵션 장난
plt.xticks(x, ['a', 'b', 'c'])
plt.yticks(sorted(y))
plt.xlabel('x axis')
plt.ylabel('y axis')

plt.show()
```

## f. matplotlib --> histogram 🏯
---
- 1차원 벡터 이상 데이터 --> hist 명령은 bins 인수로 데이터를 집계할 구간 정보를 입룍자동으로 반환
- http://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.hist

- 설정 가능한 파라미터는? = plt.hist(파라미터)
    - x : 히스토그램을 만들 데이터
    - bins : 히스토그램 구간
    - range : bins의 최소, 최대 값 튜플
    - density : True로 설정하면, 히스토그램의 합계가 1이 되도록 정규화
    - weights : 각 x의 가중치
    - cumulative : True로 설정하면, 각 바구니에 이전 바구니의 합계를 추가하여 누적 히스토그램을 생성
    - histtype : 히스토그램의 타입, 'bar', 'barstacked', 'step', 'stepfilled'
    - align : 바구니와 가로축 간의 정렬 방법, 'left', 'mid', 'right'
    - orientation : 히스토그램의 방향, 'horizontal', 'vertical'
    - color : 히스토그램의 색상
    - label : 히스토그램에 대한 레이블

```python
x = np.random.randn(1000)
plt.title("histogram")
plt.hist(x, bins=10) # arrays, bins, patches
plt.show()
```
