## d. 그럼 질문! 컴플리헨션 vs 람다, 뭘로 쓰지? 🏇
---
- 무슨차이가 있을까? = 우선 실행속력차이!

```python
import time
df = pd.DataFrame(np.random.randn(10000000*2).reshape(10000000, 2), columns=['a', 'b'])
df
```

- 람다 + map

```python
t1 = time.time()
df.assign(b = list(map(lambda x: x**2, df.a)))
t2 = time.time()
print(t2-t1)
```

- 컴프리헨션

```python
t1 = time.time()
df.assign(b = [x**2 for x in df.a])
t2 = time.time()
print(t2-t1)
```

- 결국은 취향차? = 네니오~ --> 100행까지만, a의 제곱을 다시해보자.

```python
df
```

- 람다 + apply()

```python
df.a.iloc[:100].apply(lambda x: x**2)
```

- 컴프리헨션

```python
pd.Series([a**2 for a in df.a.iloc[:100]])
```
