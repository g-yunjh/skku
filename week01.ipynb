## a. "WSL + 파워쉘 + 파이썬 가상화" + 주피터 노트북 = 직접 설치 하는 방법 💪
---
- 아나콘다 사용 X = why? 최신버전 관리는 good but 커스텀이 안됨
- WSL(window subsystem for linux) = 윈도우에서 지원하는 리눅스 서브시스템 = 리눅스 커널X
- cmd? 파워셀? = 명령줄(텍스트) 기반 인터페이스
    - cmd는 과거 버전 윈도우의 명령줄 인터페이스
    - 파워쉘은 폭넓은 기능 제공을 위해 윈도우7부터 도입된 명령줄 안터페이스
- <span style="background:yellow; color:red"> **why? WSL + 파워쉘로 해요? = 리눅스 접할 기회, 나중에 당황 말라구요**</span>
- 설치 = 최신버전 설치
    1. 윈도우 + r --> powershell
    2. $PSVersionTable # 버전보기
    3. winget search powershell # 최신버전 검색 --> 생략
    4. winget install --id Microsoft.PowerShell # 최신버전 설치끝 --> 생략
    5. 윈도우 + r --> pwsh # why not powershell? = MS 특징 --> 생략
