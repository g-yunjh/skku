## e. why? 배열 --> 행렬 계산 feat. Vectorization 🤨
---
- 아래의 히든레이어 fully connected neural network를 back-propagation 가정<br><br>
<img src="https://raw.githubusercontent.com/tysep16/DV25_1/refs/heads/main/4x4matrix_FCN.png" width="600">
  
- 결국 아래의 연립방정식 풀이 --> x, y, z, w 계산

$\begin{cases}
y+z+w = 3 \\ 
x+z+w = 3 \\ 
x+y+w = 3 \\
x+y+z = 3 
\end{cases}$

- 행렬표현?
  
$\begin{bmatrix} 
0 & 1 & 1 & 1 \\
1 & 0 & 1 & 1 \\
1 & 1 & 0 & 1 \\
1 & 1 & 1 & 0 
\end{bmatrix}
\begin{bmatrix}
x \\ 
y \\ 
z \\ 
w 
\end{bmatrix} =
\begin{bmatrix}
3 \\ 
3 \\ 
3 \\ 
3 
\end{bmatrix}$

- 풀이

$\begin{bmatrix}
x \\ 
y \\ 
z \\ 
w 
\end{bmatrix} =
\begin{bmatrix} 
0 & 1 & 1 & 1 \\
1 & 0 & 1 & 1 \\
1 & 1 & 0 & 1 \\
1 & 1 & 1 & 0 
\end{bmatrix}^{-1}
\begin{bmatrix}
3 \\ 
3 \\ 
3 \\ 
3 
\end{bmatrix}$

```python
a = np.array([[0, 1, 1, 1], [1, 0, 1, 1], [1, 1, 0, 1], [1, 1, 1, 0]])
a
```

```python
b = np.array([3, 3, 3, 3]).reshape(4,1)
b
```

```python
b = np.array([3] * 4).reshape(4, 1)
b
```

```python
b = np.array([3, 3, 3, 3]) # 얘도 가능
b
```

- How? 역행렬 계산?

```python
_a = np.arange(4).reshape(2, 2)
_a
```

```python
np.linalg.inv(_a) # np.linalg.inv는 역행렬 구해주는 메소드
```

- 즉, $A*A^{-1}$ = $E$ 단위행렬

```python
_a @ np.linalg.inv(_a) # @ = 행렬곱
```

- 다시 FCN 문제로 돌아가서...

```python
np.linalg.inv(a) @ b
```

## f. shape에 대한 유연성 = 좋아하진 않는다. 😥
---
- 엄밀하게는 아래의 행렬곱만이 가능하다. = (a, b) @ (c, d) = (a, d) --> b = c

```python
a = np.array([1, 2, 3, 4]).reshape(2, 2)
b = np.array([1, 2]).reshape(2, 1)
a@b
```

- a, b, a @ b 의 shape를 보면,

```python
a.shape, b.shape, (a@b).shape
```

- 다음과 같은 행렬 곱은 어떨까? --> 당연히 에러, why? b ~= c

```python
b = np.array([1, 2]).reshape(1, 2)
a@b
```

- but, 이건 가능 = b @ a

```python
b@a
```

```python
a.shape, b.shape, (b@a).shape
```

- 원래 항상 차원을 맞춰줘야 하지만, 아래와 같이 유연성 발휘

```python
b1 = np.array([1,2,3,4])
b2 = np.array([1,2,3,4]) 
b1@b2 
```

```python
b1.shape, b2.shape, (b1@b2).shape 
```

- (1,4) @ (4,1) = (1,1) 로 유연성 있게 처리. 원래는...

```python
b1 = np.array([1,2,3,4]).reshape(1,4) 
b2 = np.array([1,2,3,4]).reshape(4,1) 
b1@b2 
```

```python
b1.shape, b2.shape, (b1@b2).shape 
```

- 그래서 (4,1) @ (1,4)와 같은 계산은 직접 차원을 명시해야 함

```python
b1 = np.array([1,2,3,4]).reshape(4,1) 
b2 = np.array([1,2,3,4]).reshape(1,4) 
b1@b2 
```

```python
import time
import numpy as np

n = 5000
x = np.random.rand(n, n)
y = np.random.rand(n, n)

t1 = time.time()
for i in range(n):
    for j in range(n):
        z[i, j] = (x[i, j] - y[i, j])**2
t2 = time.time()
t2-t1      
```
